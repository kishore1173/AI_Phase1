# Earthquake Magnitude Prediction using LSTM


* LSTM networks are a type of recurrent neural network (RNN) capable of modeling temporal dependencies in sequential data
* LSTMs address this problem by introducing a memory cell, which is a container that can hold information for an extended period. The memory cell is controlled by three gates: the input gate, the forget gate, and the output gate.


# Import required modules

Import statements provide the necessary tools to build, train, and evaluate your deep learning model for earthquake magnitude prediction

```python
import numpy as np
import pandas as pd
import tensorflow as tf
import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
```

# Data Loading and Preprocessing

The code loads your earthquake data from a CSV file named 'resultdata.csv' using pd.read_csv.

```python
# Load your earthquake data (replace 'your_data.csv' with your dataset)
data = pd.read_csv('resultdata.csv')
```

# Feature Extraction:

* It extracts features from the dataset, which include 'Longitude', 'Latitude', 'Depth', and 'Timestamp'. These features are stored in the features array.
* The target variable, 'Magnitude', is stored in the labels array.


```python
# Assuming your dataset has columns 'longitude', 'latitude', and 'magnitude'
# Adjust this accordingly based on your actual data columns
input = data[['Longitude', 'Latitude', 'Depth','Timestamp']].values
target = data['Magnitude'].values
```

# Normalization:

The longitude and latitude features are normalized to a range between 0 and 1 using the MinMaxScaler from scikit-learn. This scaling helps neural networks perform better.

```python
# Normalize longitude and latitude features to a range between 0 and 1
scaler = MinMaxScaler()
normalized_input = scaler.fit_transform(input)
```

# Train-Test Split:
The dataset is split into training and testing sets using train_test_split from scikit-learn. The split ratio is 80% for training and 20% for testing

```python
# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    normalized_input, target, test_size=0.2, random_state=42)
```

# LSTM Model:

* Input shape: represents no of time step in the input sequence. I mention time steps in the model.
* Dropout Layer: we are adding a dropout layer to the model with a dropout rate of 0.2. at approximately 20% of neurons in the previous layer will be randomly “dropped out “ or deactivated
* Fully-Connected Layer: The output layer has a single neuron. in regression tasks where you want to predict a single continuous value, such as earthquake magnitude.
* Activation: ‘Linear’ function means that the output of the neuron is directly proportional to the weighted sum of its inputs.

```python
# Define the LSTM model
model = keras.Sequential()
model.add(keras.layers.LSTM(units=50, return_sequences=True, input_shape=(X_train.shape[1], 1)))
model.add(keras.layers.LSTM(units=50))
model.add(keras.layers.Dense(1))  # Output layer with 1 neuron for magnitude prediction
```

# Model Compilation:

The model is compiled with the 'adam' optimizer and the 'mean_squared_error' loss function. This is a regression problem, so the mean squared error is an appropriate loss function.

```python
# Compile the model
model.compile(optimizer='adam', loss='mean_squared_error')
```

# Model Training:

The model is trained using the training data with 10 epochs and a batch size of 32. It also validates the model's performance on the test data during training.

```python
model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_test, y_test))
```

# Making Predictions:

* The model is used to make predictions on the test data, and the predicted earthquake magnitudes are stored in the predictions array.
* Train the model and predict the result using the test dataset to calculate the result we use Mean Square Error.


```python


# Make predictions
predictions = model.predict(X_test)

# You can now use predictions for further analysis 
```

# Model Evaluation:

* The model evaluation is done by using evaluate()
* Evaluation is compared with predicted data with actual result 


```python
# Evaluate the model on the test set
loss = model.evaluate(X_test, y_test)
print(f'Test Loss: {loss}')
```
